1. What is a parameter?
  - Parameters are the internal values that model adjusts during training to make accurate predictions.

2. What is correlation? What does negative correlation mean?
   - Correlation describes the relationship between two variables, while negative correlation specifically refers to a relationship where one variable increases as the other decreases, and vice versa.


3. Define Machine Learning. What are the main components in Machine Learning?
   - Machine Learning is a field of computer science that enables computers to learn patterns from data and make decisions or predictions without being explicitly programmed. Main components of ML include algorithms, data, models, and the ability to make predictions.

4. How does loss value help in determining whether the model is good or not?
   - Loss value serves as a crucial indicator of a model's performance. A lower loss value suggests the model's predictions are closer to the actual values, indicating a better-performing model. Conversely, a higher loss value signifies larger discrepancies between predictions and actual values, suggesting the model needs improvement.

5. What are continuous and categorical variables?
   - continuous variables is a variable that can take any numerical value within a range and is measurable. Categorical variables is A variable that represents distinct groups or categories.

6. How do we handle categorical variables in Machine Learning? What are the common techniques?
   - Handling categorical variables in machine learning involves converting them into a numerical format that algorithms can understand. Common techniques include one-hot encoding, label encoding, binary encoding, and frequency encoding. The best technique depends on the nature of the categorical data and the specific model being used.

7. What do you mean by training and testing a dataset?
   - In machine learning, training a dataset means using a portion of the data to teach a model how to recognize patterns and make predictions. Testing a dataset involves using a separate portion of the data to evaluate the model's performance on unseen data, ensuring it generalizes well to new inputs.

8. What is sklearn.preprocessing?
   - sklearn.preprocessing module provides various functions and classes to transform raw data into a suitable format for machine learning models. This module offers tools for scaling, normalization, encoding categorical features, handling missing values, and generating polynomial features.

9. What is a Test set?
   - A test set is a portion of the dataset that is used to evaluate the performance of a trained machine learning model. It contains data that the model has never seen during training, allowing to assess how well the model generalizes to new, unseen data.

10. How do we split data for model fitting (training and testing) in Python? How do you approach a Machine Learning problem?
    - In Python, we can split data into training and testing sets using the train_test_split function from the scikit-learn library. This function randomly splits the dataset into two parts: one for training the model and the other for testing its performance.

11. Why do we have to perform EDA before fitting a model to the data?
    - Exploratory Data Analysis (EDA) should be performed before fitting a model because it helps to understand the dataset, identify potential issues like missing values, outliers, or data inconsistencies, and inform feature selection/transformation.

12. What is correlation?
    - Correlation is a statistical measure that describes the relationship between two variables. It indicates how strongly the variables are related and whether an increase in one variable is associated with an increase or decrease in the other.

13. What does negative correlation mean?
    - Negative correlation refers to a relationship where one variable increases as the other decreases, and vice versa.

14. How can you find correlation between variables in Python?
    - To find the correlation between variables in Python, we can use the Pandas library. First, create or load dataset into a DataFrame. Then, use the .corr() function to get the correlation matrix.

15. What is causation? Explain difference between correlation and causation with an example.
    - Causation means one event directly causes another, while correlation indicates a relationship between two events without necessarily implying cause and effect.

16. What is an Optimizer? What are different types of optimizers? Explain each with an example.
    - An optimizer in machine learning is an algorithm used to minimize a loss function, finding the best set of parameters (like weights and biases) for a model to make accurate predictions. The different types of optimizers are GD, SGD, Momentum, RMSProp, Adam.

17. What is sklearn.linear_model?
    - sklearn.linear_model is a module in Scikit-learn that provides linear models for regression and classification tasks.


18. What does model.fit() do? What arguments must be given?
    - model.fit() is used to train a machine learning model on your data.
It tells the model to learn patterns from the input features (X) and the target values (y). model.fit(X, y) is the arguments. X: Input features (2D array or DataFrame), y: Target/output values (1D array or Series).

19. What does model.predict() do? What arguments must be given?
    - model.predict() is used to make predictions using a trained model. The arguments is model.predict(X). X → Input features (same shape as training data, without the target)

20. What are continuous and categorical variables?
    - continuous variables represent numerical data that can take any value within a specific range, while categorical variables represent data that can be divided into distinct categories.

21. What is feature scaling? How does it help in Machine Learning?
    - Feature scaling standardizes the range of features to ensure fair contributions to the model, improving performance and speeding up learning. Common methods are Standardization (mean = 0, std = 1) and Min-Max Scaling (range 0 to 1).

22. How do we perform scaling in Python?
    - To perform feature scaling in Python, we can use the Scikit-learn library, which provides built-in classes like StandardScaler for standardization and MinMaxScaler for normalization.

23. What is sklearn.preprocessing?
    - sklearn.preprocessing is a module in Scikit-learn that provides several utilities to preprocess your data before applying machine learning algorithms. These utilities help transform or scale data into formats that are more suitable for training models.

24. How do we split data for model fitting (training and testing) in Python?
    - In Python, we can split data for model training and testing using the train_test_split function from the scikit-learn library. This function divides the data into training and testing sets, with a common split being 80% for training and 20% for testing, or similar.

25. Explain data encoding?
    - Data encoding refers to the process of converting categorical data (i.e., data with labels or categories) into a numerical format that machine learning models can process. Machine learning algorithms typically require numeric input, so encoding categorical variables is essential for model training. Types of Data Encoding are Label Encoding, One-Hot Encoding, Binary Encoding, Ordinal Encoding.
